```python
import os
import shutil
from datetime import datetime

def save_models_to_drive():
    # Dicionário mapeando as pastas locais para os seus respectivos caminhos de destino no Google Drive
    current_date = datetime.now().strftime("%Y-%m-%d")
    paths_to_copy = {
        "/content/DPED-Sistemas-Hibridos/models": os.path.join("/content/drive/MyDrive/SIH/privEmbeddings", current_date, "models"),
        "/content/DPED-Sistemas-Hibridos/results": os.path.join("/content/drive/MyDrive/SIH/privEmbeddings", current_date, "results")
    }

    for local_dir, drive_dir in paths_to_copy.items():
        if not os.path.exists(local_dir):
            print(f"Local directory '{local_dir}' does not exist yet. Skipping.")
            continue

        os.makedirs(drive_dir, exist_ok=True)
        copied_count = 0

        for root, dirs, files in os.walk(local_dir):
            for file in files:
                local_path = os.path.join(root, file)
                relative_path = os.path.relpath(local_path, local_dir)
                drive_path = os.path.join(drive_dir, relative_path)

                os.makedirs(os.path.dirname(drive_path), exist_ok=True)

                if not os.path.exists(drive_path):
                    shutil.copy2(local_path, drive_path)
                    print(f"Saved: {relative_path} -> Drive ({current_date}/{os.path.basename(local_dir)})")
                    copied_count += 1
                else:
                    print(f"Already exists on Drive: {relative_path} inside folder {current_date}/{os.path.basename(local_dir)}")

        print(f"Finished {os.path.basename(local_dir)}. Copied {copied_count} new files to Drive.")
```
